**Tarea `resolver_fecha` del job.** Parámetro `fecha`: `auto` (siguiente día FEED del calendario sin trayectorias en oro), `AAAA-MM-DD` o lista separada por comas (backfill).

**Salida.** Task value `fecha` que consumen todas las tareas siguientes; `NINGUNA` si no hay pendientes (la condición `hay_fecha_pendiente` corta el DAG).
